```python
from gtts import gTTS
text= "ㅈㄴ하고 싶은 피파 언제 끝나냐 학교 야자도 떄려쳐!! 그냥 다 떄려쳐!!!!!!."
tts = gTTS(text=text,lang="ko")

tts.save("output.mp3")
print("음성 파일 저장 완료")
```

```python
from IPython.display import Audio
Audio("output.mp3")
Audio("output.mp3",autoplay=True)
```
