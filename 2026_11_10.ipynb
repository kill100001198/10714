```python
def calculator(a,operator,b):
  if operator=='+':
      return a+b
  elif operator == '-':
      return a-b
  elif operator == '*':
      return a*b
  elif operator == '/':
      if b==0:
          return
      return a/b
  else:
      return "오류: 지원원하지 않는 연산자입니다(+,-,*,/중 하나를 입력하세요.)"
```
