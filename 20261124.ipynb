```python
def odd_even(num):
  if num %2==0:
    return '짝수'
  else:
    return '홀수'

int_num=int(input('정수를 입력하세요: '))
print(odd_even(int_num))
```

```python
def tri_area(base, height):
  area=0.5*base*height
  return area
b= int (input('삼각형의 밑변을 입력하세요:'))
h= int (input('삼각형의 높이를 입력하세요:'))
result = tri_area(b,h)
print(f'삼각형의 넓이는 {result}입니다.')
```
