```python
def op(num1,num2):
  plus = num1+num2
  minus = num1-num2
  multipy= num1*num2
  diveide= num1/num2
  return plus,minus, multipy, diveide
a,b=map(int, input('입력: ').split())
p,m, mul, d=op(a,b)
print(f'{a}+{b}={p}')
print(f'{a}-{b}={m}')
print(f'{a}*{b}={mul}')
print(f'{a}/{b}={d}')
```
